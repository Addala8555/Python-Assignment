# While Loop in Python
This notebook demonstrates how while loops work with examples.

```python
i = 0
while i < 9:
    print(i)
    i += 1
```

```python
n = 9
for i in range(n):
    if i == 0 or i == n - 1:
        print('*' * n)
    else:
        print('*' + ' ' * (n - 2) + '*')
```

```python
i = 5
while i > 0:
    print(i)
    i -= 1
```

```python
n = 3
while n > 0:
    print(n)
    n -= 1
```

```python
nums = [2, 5, 1]
i = 0
total = 0
while i < len(nums):
    total += nums[i]
    i += 1
print(total)
```
