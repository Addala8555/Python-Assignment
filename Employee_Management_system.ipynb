# Python Inheritance Practice: Employee Management System

## 1. Problem Statement

Imagine a company has different types of employees, such as Developers and Managers.

Every employee has some common information, such as:

- Employee ID
- Employee Name
- Salary

However, different employees also have their own specific information.

For example:
- A Developer works with a particular programming language, such as Python, Java, or C++.
- A Manager manages a team of employees and has a specific team size.

Instead of creating separate classes and writing the same attributes and methods repeatedly, we can use **Inheritance** in Python.

Your task is to create a parent class named `Employee` and two child classes named `Developer` and `Manager`.

Both child classes should inherit the common attributes and methods from the `Employee` class.

---

## 2. Detailed Requirements

### Requirement 1: Create a Parent Class – Employee

Create a class named `Employee`.

This class should contain the common information that every employee in the company has.

**Attributes:**

| Attribute | Description | Example |
|---|---|---|
| emp_id | Unique employee ID | 101 |
| name | Employee name | Arun |
| salary | Employee monthly salary | 50000 |

Use the `__init__()` constructor to initialize these three attributes.

**Method: `display_details()`**

- Create a method named `display_details()`.
- This method should display the employee's ID, name, and salary.
- Since these details are common to all employees, define this method only in the parent class.

### Requirement 2: Create a Child Class – Developer

Create a class named `Developer` that inherits from the `Employee` class.

A Developer is also an employee, so they should automatically get access to the attributes and methods of the parent class.

**Additional Attribute:**

| Attribute | Description | Example |
|---|---|---|
| programming_language | Programming language used by the developer | Python |

**Additional Method: `write_code()`**

- Create a method named `write_code()`.
- This method should display the programming language used by the developer.

**Important:**

- The `Developer` class should inherit from `Employee`.
- It should have access to `emp_id`, `name`, and `salary`.
- It should also have its own attribute, `programming_language`.
- You should be able to call both `display_details()` and `write_code()` using a Developer object.

### Requirement 3: Create Another Child Class – Manager

Create a class named `Manager` that inherits from the `Employee` class.

A Manager is also an employee, but instead of writing code, a manager is responsible for managing a team.

**Additional Attribute:**

| Attribute | Description | Example |
|---|---|---|
| team_size | Number of employees managed | 8 |

**Additional Method: `manage_team()`**

- Create a method named `manage_team()`.
- This method should display the number of employees managed by the manager.

**Important:**

- The `Manager` class should inherit from `Employee`.
- It should have access to `emp_id`, `name`, and `salary`.
- It should also have its own attribute, `team_size`.
- You should be able to call both `display_details()` and `manage_team()` using a Manager object.

## 6. Concepts You Should Learn

After completing this exercise, you should understand:

- How to create a parent class.
- How to create child classes.
- How to inherit attributes and methods.
- How to reuse parent class methods in child classes.
- How different child classes can have their own attributes and methods.
- How to create objects of inherited classes.

````python
class Employee:
  def __init__(self, Emp_id , Name ,Salary ):
    self.Emp_id  = Emp_id
    self.Name = Name
    self.Salary = Salary
  def display_details(self):
    print(f"Employe ID is : {self.Emp_id}")

    print (f"{self.Name}")
    print (f"{self.Salary}")

class Developer(Employee):
  def __init__(self , Emp_id , Name ,Salary ,Programming_language):
    super().__init__(Emp_id, Name, Salary)
    self.Programming_language = Programming_language

  def write_code(self):
    print (f"{self.Name} is writing code using {self.Programming_language}.")

class Manager(Employee):
  def __init__(self , Emp_id , Name ,Salary ,Team_size):
    super().__init__(Emp_id, Name, Salary)
    self.Team_size = Team_size

  def manage_team(self):
    print(f"{self.Name} is managing a team of {self.Team_size}.")

Employee1 = Developer(123 , "Surya" , 45000 , "python")
Employee2 = Manager(456, "Siva" , 100000 , 7)
````
